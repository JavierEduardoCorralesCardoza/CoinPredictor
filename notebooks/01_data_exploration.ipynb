# 01 — Data Exploration

Load BTC daily OHLCV and inspect basic structure, returns, and missing data.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / 'src'))

import pandas as pd
from coinpredictor.data.ohlcv import load_ohlcv

df = load_ohlcv()
print(df.shape, df.index.min().date(), '->', df.index.max().date())
df.tail()
```

```python
df.describe()
```

```python
# Daily returns distribution
ret = df['close'].pct_change().dropna()
ax = ret.hist(bins=80, figsize=(8, 4))
ax.set_title('BTC daily returns')
print('mean', ret.mean(), 'std', ret.std())
print('share of up-days:', (ret > 0).mean())
```

```python
# Price history
df['close'].plot(figsize=(10, 4), logy=True, title='BTC close (log scale)')
```
